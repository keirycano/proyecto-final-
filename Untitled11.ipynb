```python
import numpy as np
import matplotlib.pyplot as plt
x= np. arange (5)
y= x**2
plt.plot(x,y, "r-o" , label="lineal")
plt.plot(x,x, color ="green", linestyle= ":", marker= "s", linewidth="2", label="cuadratica")

plt.title("grafico de curvas lineal y cuadratica")
plt.xlabel("eje x")
plt.ylabel("eje y")
plt.xlim(0,6)
plt.ylim(0,6)

plt.legend()
plt.grid()
plt.show()
```

```python
plt.scatter(x,y,label= "dispersion")
plt.title("grafico de dispersion")
plt.xlabel("eje x")
plt.ylabel("eje y")
plt.legend
plt.grid()
plt.show()

```

```python
equipos=["celular" , "parlantes" , "audifonos" , "pc" , "portatil"]
cantidad= [3, 6, 2, 2, 4]
plt.bar(equipos, cantidad)
plt.title("grafico de barras")
plt.xlabel("equipos")
plt.xlabel("cantidad")
plt.grid()
plt.show()
```
